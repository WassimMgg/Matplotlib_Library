# Matplotlib example Workflow 

```python
# 0. import matplotlib and get it ready for plotting in Jupyter
import matplotlib.pyplot as plt 

# 1.prepare  data 
x =  [1, 2, 3, 4]
y = [11, 22, 33, 44]

# 2.Setup plot
fig, ax = plt.subplots(figsize=(10,10)) # Height & Width 

# 3.Plot data
ax.plot(x, y)

# 4. Customize plot
ax.set(title = "Simple Plot", 
      xlabel = "x-axis",
      ylabel = "y-axis")

# 5.Save & show (you save the whole figure)
fig.savefig("Images/sample-plot.png")
```

## Create a Bar Plot

```python
# Make a plot from dictionary
nut_butter_prices = {
    "Almond butter": 10, 
    "Peanut butter": 8,
    "Cashew butter": 12
}
fig , ax = plt.subplots()
ax.bar(nut_butter_prices.keys() ,nut_butter_prices.values())
ax.set(title = "Dan's Nut Butter Store", 
      ylabel = "Prices ($)")
```

```python
fig , ax = plt.subplots()
ax.barh(nut_butter_prices.keys(), nut_butter_prices.values())
```

## Create a Histogram

```python
# make some data for histogram
x = np.random.randn(1000) # pulls data from a normal distrubution
fig , ax = plt.subplots()
ax.hist(x)
```

## Two Options for Subplots 

```python
#Subplot option 1: Create 4 subplots with each Axes having its own variable name
fig , ((ax1, ax2) , (ax3 , ax4)) = plt.subplots(nrows = 2, # 2 rows 
                                                ncols = 2, # 2 columns 
                                               figsize = (10, 5))
# PLot to each diffrent Axis 
ax1.plot(x, x/2)
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax4.hist(np.random.randn(1000))
```

```python
# Subplots Option 2: Create 4 subplots with a single ax variable 
fig , ax = plt.subplots(nrows = 2, 
                        ncols = 2, 
                       figsize = (10, 5))
#Plot to each different index 
ax[0, 0].plot(x , x / 2)
ax[0, 1].scatter(np.random.random(10), np.random.random(10))
ax[1, 0].bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax[1, 1].hist(np.random.randn(1000))
```

# Plotting from Pandas DataFrames 

```python
import pandas as pd
```

```python
# make a DataFrame
car_sales = pd.read_csv("Datasets/car-sales.csv")
car_sales
```

```python
ts = pd.Series(np.random.randn(1000), 
              index =pd.date_range("1/1/2000" , periods=1000))
ts = ts.cumsum() # Add up the values cumulatively
ts.plot()
```

```python
car_sales["Price"] = car_sales["Price"].str.replace(r'[$,.]', '', regex=True)
car_sales
```

```python
# Remove the last two zeros
car_sales["Price"] = car_sales["Price"].str[:-2]
car_sales
```

```python
car_sales["Sale Date"] = pd.date_range('1/1/2026', periods=len(car_sales))
car_sales
```

```python
car_sales["Total Sales"]= car_sales["Price"].astype(int).cumsum() # I want them as int's not string
car_sales
```

```python
# Let's plot the total sales 
car_sales.plot(x = "Sale Date", y = "Total Sales")
```

```python
car_sales.plot(x = "Odometer (KM)" , y ="Price", kind = "scatter")
```

```python
# Convert Price to int
car_sales["Price"] = car_sales["Price"].astype(int)

# Plot a scatter plot
car_sales.plot(x="Odometer (KM)", y="Price", kind='scatter');
```
